## Key Question 2: What major seasonal patterns can be observed?

This analysis examines environmental conditions across Kharif, Rabi, and Zaid seasons to identify major seasonal patterns in rainfall, temperature, humidity, sunlight, and soil moisture.

```python
# Calculate average environmental conditions for each season

season_environment = df_clean.groupby('Season')[
    [
        'Rainfall_mm',
        'Avg_Temperature_C',
        'Humidity_pct',
        'Sunlight_Hours_Day',
        'Soil_Moisture_pct'
    ]
].mean().round(2)

season_environment
```

```python
plt.figure(figsize=(10, 6))

season_environment.plot(kind='bar', figsize=(10, 6))

plt.title('Average Environmental Conditions by Season')
plt.xlabel('Season')
plt.ylabel('Average Value')
plt.xticks(rotation=0)
plt.legend(title='Environmental Factors')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    season_environment,
    annot=True,
    fmt='.2f',
    cmap='YlGnBu'
)

plt.title('Seasonal Environmental Conditions Heatmap')
plt.xlabel('Environmental Factors')
plt.ylabel('Season')

plt.show()
```

### Observation

The environmental conditions show clear seasonal variations.

- Kharif receives the highest average rainfall of 849.20 mm and has the highest average humidity of 71.81%.
- Kharif also has the highest average soil moisture at 31.15%.
- Rabi has the lowest average temperature at 23.49°C.
- Zaid receives the lowest average rainfall of 304.65 mm.
- Zaid has the highest average temperature of 31.04°C and the lowest average humidity of 52.01%.
- Zaid also records the highest average sunlight exposure at 8.18 hours per day.
- Soil moisture decreases from Kharif to Rabi and is lowest in Zaid.

### Insight

The major seasonal pattern is a shift from wetter and more humid conditions in Kharif to drier and hotter conditions in Zaid. These environmental differences provide useful context for understanding the seasonal differences in agricultural performance.

### Conclusion

Agricultural seasons have noticeably different environmental conditions. Rainfall, temperature, humidity, sunlight, and soil moisture all vary across seasons and should be considered when evaluating agricultural performance and planning seasonal farming activities.

### Observation

The distribution of crops and irrigation methods varies across the three seasons.

- Rice has the highest number of farms in all three seasons, with 314 farms in Kharif, 274 in Rabi, and 102 in Zaid.
- Maize and Wheat also have relatively high representation, particularly during Kharif and Rabi.
- Sugarcane has the lowest representation in Kharif and Zaid, while its count is slightly higher in Rabi.
- Overall, Kharif has more crop records than Rabi and Zaid.
- Flood irrigation is the most common irrigation method in all three seasons.
- Rainfed irrigation is the second most common method in Kharif and Rabi, while Drip and Sprinkler have lower representation.
- All irrigation methods have substantially fewer records in Zaid compared with Kharif and Rabi.

### Insight

The data shows that agricultural activities are not distributed equally across seasons. Crop participation and irrigation practices differ between seasons, with Rice being consistently prominent and Flood irrigation being the most commonly observed irrigation method.

### Conclusion

Agricultural activities vary across seasons in terms of crop distribution and irrigation methods. These differences should be considered when evaluating seasonal agricultural performance and resource requirements.

## Key Question 6: Are there relationships between seasonal environmental conditions and agricultural performance?

This analysis investigates the relationships between environmental conditions and agricultural performance using correlation analysis. Environmental variables such as rainfall, temperature, humidity, sunlight, soil pH, and soil moisture are compared with yield and production.

```python
environment_performance = df_clean[
    [
        'Rainfall_mm',
        'Avg_Temperature_C',
        'Humidity_pct',
        'Sunlight_Hours_Day',
        'Soil_pH',
        'Soil_Moisture_pct',
        'Yield_Tonnes_Ha',
        'Production_Tonnes'
    ]
]

correlation_matrix = environment_performance.corr().round(2)

correlation_matrix
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Between Environmental Conditions and Agricultural Performance')
plt.show()
```

### Observation

The correlation analysis shows that the environmental variables have very weak linear relationships with agricultural yield in this dataset.

- Rainfall has a correlation of 0.03 with yield.
- Average temperature has a correlation of 0.01 with yield.
- Humidity has a correlation of 0.01 with yield.
- Sunlight hours have a correlation of -0.02 with yield.
- Soil pH has a correlation of -0.02 with yield.
- Soil moisture has a correlation of 0.01 with yield.

Some relationships are more noticeable among the environmental variables themselves. Rainfall has a moderate positive correlation with humidity (0.52) and soil moisture (0.51), while humidity and soil moisture have a positive correlation of 0.45.

Yield and production have a strong positive correlation of 0.88.

### Insight

The available data does not show a strong linear relationship between individual environmental variables and yield. However, environmental variables are related to each other, particularly rainfall, humidity, and soil moisture.

### Conclusion

Based on correlation analysis, no strong linear relationship between the individual environmental conditions and yield can be established from this dataset. This does not mean that environmental conditions have no influence on agriculture; other factors, combinations of variables, crop type, farming practices, and seasonal conditions may also contribute to agricultural performance.

# Limitations

The analysis has some limitations that should be considered while interpreting the findings:

1. **Correlation does not imply causation**  
   The correlation analysis identifies relationships between variables but does not prove that one variable directly causes changes in another.

2. **Limited environmental variables**  
   The dataset contains selected environmental conditions. Other factors that may influence agricultural performance may not be included.

3. **Historical dataset limitations**  
   The analysis is based only on the available dataset and therefore the findings may not represent all agricultural regions or farming conditions.

4. **Extreme observations**  
   Some variables contain unusually high or low observations. These values may influence averages and should be investigated further.

5. **Regional differences**  
   Agricultural conditions can vary considerably between states and locations. The available data may not capture every local factor.

6. **Economic factors**  
   Market prices, costs, and profits can change over time. Therefore, economic findings should be interpreted in the context of the available data.

### Summary of Limitation

The findings provide useful evidence about patterns and differences within the available dataset, but they should not be treated as universal conclusions for all agricultural situations.
